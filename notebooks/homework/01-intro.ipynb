```python
horsepower_median = df["horsepower"].median()
```

```python
horsepower_mode = df["horsepower"].mode()[0]
print(horsepower_mode)
```

```python
df["horsepower"] = df["horsepower"].fillna(value=horsepower_mode)
```

```python
horsepower_na_filled_median = df["horsepower"].median()
```

```python
print("Before na was remove median: " + str(horsepower_median))
print("After na was removed median: " + str(horsepower_na_filled_median))
```

```python
# Confirm if the number of n/a column has reduced after filling horsepower n/a with the observed mode
df.isna().any().sum()
```

```python
# Asian cars
df_asian_cars = df[df["origin"] == "Asia"]
```

```python
df_asian_cars = df_asian_cars[["vehicle_weight", "model_year"]]
# select first serven records
X = df_asian_cars[:7].to_numpy()
```

```python
X
```

```python
XT = X.transpose()
```

```python
XT
```

```python
XTX = np.matmul(XT, X)
XTX
```

```python
# Inverse XTX. using pseudo inverse as normal inverse throws error being a non singular matrix
XTX_inv = np.linalg.pinv(XTX)
```

```python
print("X:", X.shape)
print("XT:", XT.shape)
print("XTX:", XTX.shape)
print("XTX_inv:", XTX_inv.shape)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
# @ is like a short form of np.matmul
w = np.matmul(XTX_inv, XT) @ y
```
